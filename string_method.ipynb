# 1- Upper() -> it converts text to uppercase

```python
a = "shantanu"
print(a)
print(a.upper())
print("-"*100)
b = "boo yeah"
print(b)
print(b.upper())
print("-"*100)
c = "lets fucking go"
print(c)
print(c.upper())
print("-"*100)
d = "my name is john cenaaaa!"
print(d)
print(d.upper())
print("-"*100)
e="y2j"
print(e)
print(e.upper())        
print("-"*100)    
```

# 2- Lower() -> it converts text to lowercase

```python
a = "SHANTANU"
print(a)
print(a.lower())
print("-"*100)
b = "BOO YEAH"
print(b)
print(b.lower())
print("-"*100)
c = "LETS FUCKING GO"
print(c)
print(c.lower())
print("-"*100)
d = "MY NAME IS JOHN CENAAAA!"
print(d)
print(d.lower())
print("-"*100)
e="Y2J"
print(e)
print(e.lower())        
print("-"*100)
```

# 5-Strip() -> remove spaces from start to end

```python
a = "shantanu  "
print(a)
print(a.strip())
print("-"*100)
b = " boo yeah"
print(b)
print(b.strip().upper())
print("-"*100)
c = "lets   fucking go"
print(c)
print(c.strip().upper())
print("-"*100)
d = "          my            name is john cenaaaa!"
print(d)
print(d.strip().upper())
print("-"*100)
e="y 2 j"
print(e)
print(e.strip().upper())        
print("-"*100)
```

# 6-Replace(old,new) -> replace text


```python
a = "shantanu"
print(a)
print(a.replace("s","rr"))
print("-"*100)
b = "boo yeah"
print(b)
print(b.replace("boo","hell"))
print("-"*100)
c = "lets fucking go"
print(c)
print(c.replace("go","go !"))
print("-"*100)
d = "my name is john cenaaaa!"
print(d)
print(d.replace("my","MY"))
print("-"*100)
e="y2j"
print(e)
print(e.replace("y2j","HBK"))        
print("-"*100)
```

# 8-find() -> it find position of that word

```python
a = "shantanu"
print(a)
print(a.find("s"))
print("-"*100)
b = "boo yeah"
print(b)
print(b.find("boo"))
print("-"*100)
c = "lets fucking go"
print(c)
print(c.find("go"))
print("-"*100)
d = "my name is john cenaaaa!"
print(d)
print(d.find("john"))
print("-"*100)
e="y2j"
print(e)
print(e.find("y2j"))        
print("-"*100)
```

# 9-count() -> it counts how many times that word appear

```python
a = "shantanu"
print(a)
print(a.count("s"))
print("-"*100)
b = "boo yeah"
print(b)
print(b.count("o"))
print("-"*100)
c = "lets fucking go"
print(c)
print(c.count("lets"))
print("-"*100)
d = "my name is john cenaaaa!"
print(d)
print(d.count("a"))
print("-"*100)
e="y2j"
print(e)
print(e.count("y2j"))        
print("-"*100)
```

# 10- startwith()/endwith() - checks if a string starts or end with given text

```python
a = "shantanu"
print(a)
print(a.startswith("s"))
print("-"*100)
b = "boo yeah"
print(b)
print(b.startswith("y"))
print("-"*100)
c = "lets fucking go"
print(c)
print(c.endswith("go"))
print("-"*100)
d = "my name is john cenaaaa!"
print(d)
print(d.endswith("ohn"))
print("-"*100)
e="y2j"
print(e)
print(e.endswith("j"))        
print("-"*100)    
```
