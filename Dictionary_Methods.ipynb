# 1. get() -->> it returns the value of specified key

```python
student = {
    "id": 100,
    "name": "Shantanu",
    "city": "nagpur",
    "marks": 88,
    "skills": ["python", "sql"]
}
print(student.get("name"))
print(student.get("id"))
print(student.get("country","Key does not exist"))
print(f"city of student = {student.get("city")}, length of city={len(student.get("city"))}, type = {type(student.get("city"))}")
```

# 2. keys() -->> it returns a list containing dictionary's keys

```python
emp = {
    "id": 100,
    "name": "Shantanu",
    "dept": "IT",
    "salary": 50000,
    "city": "Nagpur"
}

print(emp.keys())
s=emp.keys()
print(f"s = {s}")
emp["country"]="INDIA"
print(s)
```

# 3. values() -->> it returns a list of all the values in the dictionaries

```python
car = {
    "brand": "Tata",
    "model": "Nexon",
    "year": 2026,
    "price": 1500000,
    "colors": ["red", "black"]
}
print(car.values())
car["fuel_type"]=["petrol","diesel","ev"]
print(car)
print(car.keys())
print(car.values())
```

# 4. items() -->> it returns a list containing a tuple for each key value pair
#         -->> tuples stores multiple items in variable, it is ordered and unchangeable and can have duplicates are indexed and in () round brackets

```python
book = {
    "title": "Python",
    "author": "ABC",
    "pages": 300,
    "price": 450,
    "available": True
}

print(book.items())
book["read"]="not yet"
print(book)
print(f"book = {book.items()}")
```

# 5. update() -->> it updates the dictionary with the specified key value pair

```python
user = {
    "id": 7,
    "name": "Shantanu Sunil Sorakhe",
    "age": 25,
    "city": "Mumbai",
    "skills": ["sql","power bi","ms excel", "Python"]
}

user.update({"age": 26, "city": "Pune"})

print(user)
user.update({"age": 0, "city": "Ngpur"})
print(user)
```

# 6. len() -->> length of dictionary


```python
d = {"a": 1, "b": 2, "c": 3, "d":55}
print(f"length od dictionary d = {len(d)},type={type(d)}")
d["e"]=69
print(f"length od dictionary d = {len(d)},type={type(d)}") 
```

# 7. in() -->>  it checks if key exist

```python
d = {"name": "ss", "age": 25,"skill":["sql","python"]}
print("name" in d)
print("game" in d)
```

# 8. del() -->> it delets key

```python
d = {"a": 1, "b": 2,"name":107,"stable":"highly unstable"}
del d["a"]
print(d)
```

# 9. sorted() -->> it sort keys


```python
d = {"c": 3, "a": 1, "b": 2,"ss":84848,"aa":5498494}
print(sorted(d))
```

# 10. pop() -->> itt remove specific keys

```python
d = {
    "name": "shantanu",
    "age": 25,
    "city": "Pune",
    "job": "Engineer",
    "salary": 50000,
    "status": "Single"
}

d.pop("city")

print(d)
```

# 11. popitem() -->> it removes last inserted item

```python
d = {
    "name": "shantanu",
    "age": 25,
    "city": "Pune",
    "job": "Engineer",
    "salary": 50000,
    "status": "Single"
}

d.popitem()
d["married"]="yes"

print(d)
d.popitem()
print(d)
```
