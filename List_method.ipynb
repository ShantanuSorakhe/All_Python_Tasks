# 1. append -> aadds items at end of list

```python
a=[1,2,3,4,5]
a.append(6)
print(f"a={a}")   

b = ["a", "b"]
b.append("c")
print(f"b={b}")  

c = [10]
c.append(20)
print(f"c={c}") 
```

# 2. extend -> adds mulitple element to lists


```python
a = [1,2]
a.extend([3,4])
print(a)

a = ["x"]
a.extend(["y","z"])
print(a)

a = [1]
a.extend([2,3,4])
print(a)
```

# 3. insert -> adds element at specific position


```python
a = [1,2,3]
a.insert(1,100)
print(a)

a = ["a","b"]
a.insert(0,"x")
print(a)

a = [10,20]
a.insert(2,30)
print(a)
```

# 4.remove() -> it removes first matching value


```python
a=[1,2,3]
a.remove(2)
print(a)


a=["a","b","c"]
a.remove("b")
print(a)


a=[10,20,30]
a.remove(10)
print(a)


a=[5,5,6]
a.remove(5)
print(a)


a=[1,2,3,2]
a.remove(2)
print(a)
```

# 5. pop() -> it removes element using index

```python
a=[1,2,3]
a.pop()
print(a)


a=[1,2,3]
a.pop(0)
print(a)


a=["x","y","z"]
a.pop(1)
print(a)


a=[10,20]
a.pop()
print(a)


a=[5]
a.pop()
print(a)
```

# clear() -> it removes all elements

```python
a=[1,2]
a.clear()
print(a)


a=["a"]
a.clear()
print(a)


a=[10,20,30]
a.clear()
print(a)


a=[5]
a.clear()
print(a)


a=[1,2,3,4]
a.clear()
print(a)
```
